#  Q2 mathematically closer  numbers in array to X 

```python

arr = np.random.randint(1, 101, 10)
print(arr)

x = int(input("Enter X: "))

closest = arr[0]

for num in arr:
    if abs(num - x) < abs(closest - x):
        closest = num

print("Closest value:", closest)
```
